### 1. Melhorando o ajuste no percentual de gorjetas

a. Vamos partir do modelo feito na última tarefa, o que relaciona ```tip_pct``` e ```net_bill```. Carregue a base, os pacotes e reproduza este modelo aqui.

````python
import pandas as pd
import numpy as np
import seaborn as sns
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt

print("--- Exercício 2 - Melhorando o Ajuste no Percentual de Gorjetas ---")
print("a. Reproduzindo o modelo: tip_pct por net_bill")

# 1. Carregar a base de dados de gorjetas
try:
    tips = sns.load_dataset("tips")
    print("Base de dados 'tips' do Seaborn carregada com sucesso!")
except Exception as e:
    print(f"Erro ao carregar a base 'tips' do Seaborn: {e}")
    try:
        tips = pd.read_csv('tips.csv') # Alternativa se não carregar do seaborn
        print("Base de dados 'tips.csv' carregada com sucesso!")
    except FileNotFoundError:
        print("Erro: 'tips.csv' não encontrado. Por favor, forneça o arquivo da base de gorjetas.")
        exit()

# 2. Criar a variável 'net_bill' (conta líquida)
tips['net_bill'] = tips['total_bill'] - tips['tip']
print("\nVariável 'net_bill' criada: total_bill - tip")

# 3. Criar a variável 'tip_pct' (porcentagem da gorjeta)
# Verificar e remover net_bill = 0 para evitar divisão por zero
if (tips['net_bill'] == 0).any():
    print("\nAviso: Há 'net_bill' igual a zero. Removendo essas linhas para evitar divisão por zero em 'tip_pct'.")
    tips = tips[tips['net_bill'] != 0].copy()

tips['tip_pct'] = tips['tip'] / tips['net_bill']
print("Variável 'tip_pct' criada: tip / net_bill")

print("\nPrimeiras 5 linhas da base de dados com as variáveis criadas:")
print(tips[['total_bill', 'tip', 'net_bill', 'tip_pct']].head())

# 4. Ajustar o modelo de regressão linear: tip_pct ~ net_bill
print("\nRealizando Regressão Linear: tip_pct ~ net_bill")
model_tip_pct = smf.ols('tip_pct ~ net_bill', data=tips).fit()

# Exibir o sumário do modelo
print(model_tip_pct.summary())

print("\n--- Modelo 'tip_pct ~ net_bill' reproduzido com sucesso! ---")
````

### 2. ajuste o modelo de ```tip_pct``` em função do logaritmo de ```net_bill```.

````python
import pandas as pd
import numpy as np
import seaborn as sns
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt

print("--- Exercício 2.b. Ajustando Modelo: tip_pct por log(net_bill) ---")

# Re-carregar a base de dados e criar 'net_bill' e 'tip_pct' para garantir o ambiente
try:
    tips = sns.load_dataset("tips")
except Exception as e:
    print(f"Erro ao carregar a base 'tips' do Seaborn: {e}")
    try:
        tips = pd.read_csv('tips.csv')
    except FileNotFoundError:
        print("Erro: 'tips.csv' não encontrado. Por favor, forneça o arquivo da base de gorjetas.")
        exit()

tips['net_bill'] = tips['total_bill'] - tips['tip']

# Tratar casos de net_bill = 0 para evitar divisão por zero na tip_pct e log(net_bill)
# Também remover valores <= 0 para o logaritmo
if (tips['net_bill'] <= 0).any():
    print("\nAviso: Há 'net_bill' menor ou igual a zero. Removendo essas linhas para evitar problemas com 'tip_pct' e log().")
    tips = tips[tips['net_bill'] > 0].copy()

tips['tip_pct'] = tips['tip'] / tips['net_bill']

# --- Novo Passo: Criar a variável 'log_net_bill' ---
# Usamos np.log para o logaritmo natural
tips['log_net_bill'] = np.log(tips['net_bill'])
print("\nVariável 'log_net_bill' criada: log(net_bill)")
print("Primeiras 5 linhas da base de dados com 'log_net_bill':")
print(tips[['net_bill', 'log_net_bill', 'tip_pct']].head())

# --- Ajustar o Modelo de Regressão Linear: tip_pct ~ log_net_bill ---
print("\nRealizando Regressão Linear: tip_pct ~ log_net_bill")
model_log_net_bill = smf.ols('tip_pct ~ log_net_bill', data=tips).fit()

# Exibir o sumário do modelo
print(model_log_net_bill.summary())

print("\n--- Modelo 'tip_pct ~ log_net_bill' ajustado com sucesso! ---")
print("Observe o R-quadrado e os coeficientes deste novo modelo em comparação com o anterior.")
````

###  3. ajuste o modelo de ```tip_pct``` em função de um polinômio de segundo grau em ```net_bill```.

````python
import pandas as pd
import numpy as np
import seaborn as sns
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt

print("--- Exercício 2.c. Ajustando Modelo: tip_pct por Polinômio de 2º Grau em net_bill ---")

# Re-carregar a base de dados e criar 'net_bill' e 'tip_pct' para garantir o ambiente
try:
    tips = sns.load_dataset("tips")
except Exception as e:
    print(f"Erro ao carregar a base 'tips' do Seaborn: {e}")
    try:
        tips = pd.read_csv('tips.csv')
    except FileNotFoundError:
        print("Erro: 'tips.csv' não encontrado. Por favor, forneça o arquivo da base de gorjetas.")
        exit()

tips['net_bill'] = tips['total_bill'] - tips['tip']

# Tratar casos de net_bill = 0 para evitar divisão por zero na tip_pct
if (tips['net_bill'] == 0).any():
    print("\nAviso: Há 'net_bill' igual a zero. Removendo essas linhas para evitar problemas com 'tip_pct'.")
    tips = tips[tips['net_bill'] != 0].copy()

tips['tip_pct'] = tips['tip'] / tips['net_bill']

# --- Novo Passo: Criar a variável 'net_bill_sq' (net_bill ao quadrado) ---
tips['net_bill_sq'] = tips['net_bill']**2
print("\nVariável 'net_bill_sq' criada: net_bill^2")
print("Primeiras 5 linhas da base de dados com 'net_bill' e 'net_bill_sq':")
print(tips[['net_bill', 'net_bill_sq', 'tip_pct']].head())

# --- Ajustar o Modelo de Regressão Linear: tip_pct ~ net_bill + net_bill_sq ---
# A fórmula em statsmodels permite usar o operador I() para expressões matemáticas
# ou simplesmente passar as variáveis já criadas.
# Usando as variáveis criadas diretamente para maior clareza aqui.
print("\nRealizando Regressão Linear: tip_pct ~ net_bill + net_bill_sq")
model_poly2 = smf.ols('tip_pct ~ net_bill + net_bill_sq', data=tips).fit()

# Exibir o sumário do modelo
print(model_poly2.summary())

print("\n--- Modelo 'tip_pct ~ net_bill + net_bill^2' ajustado com sucesso! ---")
print("Observe o R-quadrado e os coeficientes deste novo modelo em comparação com os anteriores.")
````

###  4. ajuste um modelo no log do ```tip_pct``` em função de ```net_bill```. Calcule o $R^2$ também em ```tip_pct```

````python
import pandas as pd
import numpy as np
import seaborn as sns
import statsmodels.formula.api as smf
from sklearn.metrics import r2_score # Para calcular R2

print("--- Exercício 2.d. Ajustando Modelo: log(tip_pct) por net_bill ---")

# Re-carregar a base de dados e criar 'net_bill' e 'tip_pct' para garantir o ambiente
try:
    tips = sns.load_dataset("tips")
except Exception as e:
    print(f"Erro ao carregar a base 'tips' do Seaborn: {e}")
    try:
        tips = pd.read_csv('tips.csv')
    except FileNotFoundError:
        print("Erro: 'tips.csv' não encontrado. Por favor, forneça o arquivo da base de gorjetas.")
        exit()

tips['net_bill'] = tips['total_bill'] - tips['tip']

# Tratar casos de net_bill = 0 para evitar divisão por zero na tip_pct
if (tips['net_bill'] <= 0).any():
    print("\nAviso: Há 'net_bill' menor ou igual a zero. Removendo essas linhas para evitar problemas com 'tip_pct' e log().")
    tips = tips[tips['net_bill'] > 0].copy()

tips['tip_pct'] = tips['tip'] / tips['net_bill']

# --- Novo Passo: Tratar e Criar a variável 'log_tip_pct' ---
# Garantir que tip_pct seja estritamente positivo para aplicar logaritmo
if (tips['tip_pct'] <= 0).any():
    print("\nAviso: Há 'tip_pct' menor ou igual a zero. Removendo essas linhas para aplicar logaritmo.")
    tips = tips[tips['tip_pct'] > 0].copy()

tips['log_tip_pct'] = np.log(tips['tip_pct'])
print("\nVariável 'log_tip_pct' criada: log(tip_pct)")
print("Primeiras 5 linhas da base de dados com 'net_bill' e 'log_tip_pct':")
print(tips[['net_bill', 'tip_pct', 'log_tip_pct']].head())

# --- Ajustar o Modelo de Regressão Linear: log_tip_pct ~ net_bill ---
print("\nRealizando Regressão Linear: log_tip_pct ~ net_bill")
model_log_tip_pct = smf.ols('log_tip_pct ~ net_bill', data=tips).fit()

# Exibir o sumário do modelo (R2 deste modelo é para log_tip_pct)
print(model_log_tip_pct.summary())

# --- Calcular o R2 também em tip_pct ---
# 1. Obter os valores preditos do modelo na escala logarítmica
tips['log_tip_pct_pred'] = model_log_tip_pct.fittedvalues

# 2. Transformar os valores preditos de volta para a escala original (tip_pct) usando np.exp()
tips['tip_pct_pred_from_log_model'] = np.exp(tips['log_tip_pct_pred'])
print("\nValores preditos para 'tip_pct' obtidos e transformados de volta da escala logarítmica.")
print("Primeiras 5 linhas com 'tip_pct' real e predito:")
print(tips[['tip_pct', 'tip_pct_pred_from_log_model', 'net_bill']].head())

# 3. Calcular o R2 comparando 'tip_pct' real com 'tip_pct_pred_from_log_model'
r2_tip_pct_from_log_model = r2_score(tips['tip_pct'], tips['tip_pct_pred_from_log_model'])
print(f"\nR-quadrado (R²) em 'tip_pct' (predito a partir do modelo log_tip_pct ~ net_bill): {r2_tip_pct_from_log_model:.4f}")

print("\n--- Modelo 'log(tip_pct) ~ net_bill' ajustado e R-quadrado em 'tip_pct' calculado! ---")
print("Este R-quadrado é diretamente comparável com o R-quadrado do modelo 'tip_pct ~ net_bill' do item 2.a.")
````

###  5. ajuste um modelo no log do ```tip_pct``` em função do log de ```net_bill```. Calcule o $R^2$ também em ```tip_pct```.

````python
import pandas as pd
import numpy as np
import seaborn as sns
import statsmodels.formula.api as smf
from sklearn.metrics import r2_score # Para calcular R2

print("--- Exercício 2.e. Ajustando Modelo: log(tip_pct) por log(net_bill) ---")

# Re-carregar a base de dados e criar 'net_bill' e 'tip_pct' para garantir o ambiente
try:
    tips = sns.load_dataset("tips")
except Exception as e:
    print(f"Erro ao carregar a base 'tips' do Seaborn: {e}")
    try:
        tips = pd.read_csv('tips.csv')
    except FileNotFoundError:
        print("Erro: 'tips.csv' não encontrado. Por favor, forneça o arquivo da base de gorjetas.")
        exit()

tips['net_bill'] = tips['total_bill'] - tips['tip']

# --- Tratar net_bill e tip_pct para garantir valores positivos para o logaritmo ---
# Remover linhas onde net_bill é <= 0
if (tips['net_bill'] <= 0).any():
    print("\nAviso: Há 'net_bill' menor ou igual a zero. Removendo essas linhas.")
    tips = tips[tips['net_bill'] > 0].copy()

tips['tip_pct'] = tips['tip'] / tips['net_bill']

# Remover linhas onde tip_pct é <= 0
if (tips['tip_pct'] <= 0).any():
    print("\nAviso: Há 'tip_pct' menor ou igual a zero. Removendo essas linhas.")
    tips = tips[tips['tip_pct'] > 0].copy()

# --- Criar as variáveis 'log_net_bill' e 'log_tip_pct' ---
tips['log_net_bill'] = np.log(tips['net_bill'])
tips['log_tip_pct'] = np.log(tips['tip_pct'])
print("\nVariáveis 'log_net_bill' e 'log_tip_pct' criadas.")
print("Primeiras 5 linhas da base de dados com as novas variáveis:")
print(tips[['net_bill', 'log_net_bill', 'tip_pct', 'log_tip_pct']].head())

# --- Ajustar o Modelo de Regressão Linear: log_tip_pct ~ log_net_bill ---
print("\nRealizando Regressão Linear: log_tip_pct ~ log_net_bill")
model_log_log = smf.ols('log_tip_pct ~ log_net_bill', data=tips).fit()

# Exibir o sumário do modelo (R2 deste modelo é para log_tip_pct)
print(model_log_log.summary())

# --- Calcular o R2 também em tip_pct ---
# 1. Obter os valores preditos do modelo na escala logarítmica
tips['log_tip_pct_pred_from_log_log_model'] = model_log_log.fittedvalues

# 2. Transformar os valores preditos de volta para a escala original (tip_pct) usando np.exp()
tips['tip_pct_pred_from_log_log_model'] = np.exp(tips['log_tip_pct_pred_from_log_log_model'])
print("\nValores preditos para 'tip_pct' obtidos e transformados de volta da escala logarítmica.")
print("Primeiras 5 linhas com 'tip_pct' real e predito:")
print(tips[['tip_pct', 'tip_pct_pred_from_log_log_model', 'net_bill']].head())

# 3. Calcular o R2 comparando 'tip_pct' real com 'tip_pct_pred_from_log_log_model'
r2_tip_pct_from_log_log_model = r2_score(tips['tip_pct'], tips['tip_pct_pred_from_log_log_model'])
print(f"\nR-quadrado (R²) em 'tip_pct' (predito a partir do modelo log(tip_pct) ~ log(net_bill)): {r2_tip_pct_from_log_log_model:.4f}")

print("\n--- Modelo 'log(tip_pct) ~ log(net_bill)' ajustado e R-quadrado em 'tip_pct' calculado! ---")
print("Este R-quadrado é diretamente comparável com os R-quadrados dos modelos 'tip_pct ~ net_bill' (item 2.a) e 'log(tip_pct) ~ net_bill' (item 2.d).")
````

###  6. qual modelo resultou no melhor ajuste?

````python
import pandas as pd
import numpy as np
import seaborn as sns
import statsmodels.formula.api as smf
from sklearn.metrics import r2_score

print("--- Comparando R-quadrado dos Modelos para 'tip_pct' ---")

# --- Carregar e preparar os dados (bloco completo para reprodutibilidade) ---
try:
    tips = sns.load_dataset("tips")
except Exception as e:
    print(f"Erro ao carregar a base 'tips' do Seaborn: {e}")
    try:
        tips = pd.read_csv('tips.csv')
    except FileNotFoundError:
        print("Erro: 'tips.csv' não encontrado. Por favor, forneça o arquivo da base de gorjetas.")
        exit()

tips['net_bill'] = tips['total_bill'] - tips['tip']

# Tratar valores problemáticos para log e tip_pct
initial_rows = len(tips)
tips = tips[tips['net_bill'] > 0].copy()
tips['tip_pct'] = tips['tip'] / tips['net_bill']
tips = tips[tips['tip_pct'] > 0].copy() # Para log(tip_pct)
print(f"Número de linhas após limpeza de valores <= 0 em net_bill/tip_pct: {len(tips)} (original: {initial_rows})")

# Criar variáveis transformadas
tips['log_net_bill'] = np.log(tips['net_bill'])
tips['net_bill_sq'] = tips['net_bill']**2
tips['log_tip_pct'] = np.log(tips['tip_pct'])
# --- Fim da preparação dos dados ---


# --- Modelos e cálculo do R-quadrado ---

# Modelo 2.a: tip_pct ~ net_bill
model_2a = smf.ols('tip_pct ~ net_bill', data=tips).fit()
r2_2a = model_2a.rsquared
print(f"\nModelo 2.a (tip_pct ~ net_bill): R2 = {r2_2a:.4f}")

# Modelo 2.b: tip_pct ~ log_net_bill
model_2b = smf.ols('tip_pct ~ log_net_bill', data=tips).fit()
r2_2b = model_2b.rsquared
print(f"Modelo 2.b (tip_pct ~ log_net_bill): R2 = {r2_2b:.4f}")

# Modelo 2.c: tip_pct ~ net_bill + net_bill_sq (Polinômio de 2º grau)
model_2c = smf.ols('tip_pct ~ net_bill + net_bill_sq', data=tips).fit()
r2_2c = model_2c.rsquared
print(f"Modelo 2.c (tip_pct ~ net_bill + net_bill_sq): R2 = {r2_2c:.4f}")

# Modelo 2.d: log_tip_pct ~ net_bill (R2 calculado em tip_pct)
model_2d = smf.ols('log_tip_pct ~ net_bill', data=tips).fit()
# Previsões em escala logarítmica
tips['log_tip_pct_pred_2d'] = model_2d.fittedvalues
# Transformar de volta para tip_pct
tips['tip_pct_pred_2d'] = np.exp(tips['log_tip_pct_pred_2d'])
r2_2d = r2_score(tips['tip_pct'], tips['tip_pct_pred_2d'])
print(f"Modelo 2.d (log_tip_pct ~ net_bill, R2 em tip_pct): {r2_2d:.4f}")

# Modelo 2.e: log_tip_pct ~ log_net_bill (R2 calculado em tip_pct)
model_2e = smf.ols('log_tip_pct ~ log_net_bill', data=tips).fit()
# Previsões em escala logarítmica
tips['log_tip_pct_pred_2e'] = model_2e.fittedvalues
# Transformar de volta para tip_pct
tips['tip_pct_pred_2e'] = np.exp(tips['log_tip_pct_pred_2e'])
r2_2e = r2_score(tips['tip_pct'], tips['tip_pct_pred_2e'])
print(f"Modelo 2.e (log_tip_pct ~ log_net_bill, R2 em tip_pct): {r2_2e:.4f}")

# --- Comparação Final ---
r2_values = {
    '2.a (tip_pct ~ net_bill)': r2_2a,
    '2.b (tip_pct ~ log_net_bill)': r2_2b,
    '2.c (tip_pct ~ net_bill + net_bill_sq)': r2_2c,
    '2.d (log_tip_pct ~ net_bill, R2 em tip_pct)': r2_2d,
    '2.e (log_tip_pct ~ log_net_bill, R2 em tip_pct)': r2_2e
}

best_model_name = max(r2_values, key=r2_values.get)
best_r2 = r2_values[best_model_name]

print("\n--- Resultado da Comparação ---")
print(f"O modelo que resultou no melhor ajuste (maior R-quadrado para 'tip_pct') é: ")
print(f"  **{best_model_name}** com R-quadrado de **{best_r2:.4f}**")

print("\n--- Comentário ---")
print("O modelo com o maior R-quadrado para a variável `tip_pct` é considerado o que melhor explica a variância da porcentagem da gorjeta. Transformações logarítmicas e polinômiais são frequentemente eficazes para capturar relações não lineares e lidar com heterocedasticidade, levando a um ajuste melhor.")
````
